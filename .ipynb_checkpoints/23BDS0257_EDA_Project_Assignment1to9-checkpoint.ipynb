```python
# Question: Create three dataframes named dataframe1, dataframe2, and dataframe3.

import pandas as pd
import numpy as np
```

```python
# Question: Create dataframe1 containing ProductID, NameofProd, and Price.

dataframe1 = pd.DataFrame({
    "ProductID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "NameofProd": ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer",
                   "Tablet", "Webcam", "Speaker", "Headphones", "Router"],
    "Price": [55000, 800, 1500, 12000, 8500, 22000, 2500, 3000, 1800, 2200]
})

display(dataframe1)
```

```python
# Question: Create dataframe2 containing ProductID and Company_Location.

dataframe2 = pd.DataFrame({
    "ProductID": [101, 102, 103, 104, 105, 111, 112, 113, 114, 115],
    "Company_Location": ["Chennai", "Mumbai", "Delhi", "Bengaluru", "Hyderabad",
                         "Pune", "Kolkata", "Jaipur", "Kochi", "Ahmedabad"]
})

display(dataframe2)
```

```python
# Question: Create dataframe3 containing ProductID, NameofProd, and Price.

dataframe3 = pd.DataFrame({
    "ProductID": [111, 112, 113, 114, 115, 116, 117, 118, 119, 120],
    "NameofProd": ["Smartwatch", "Power Bank", "Microphone", "Projector",
                   "External Drive", "USB Cable", "Graphics Card",
                   "SSD", "RAM", "Laptop Stand"],
    "Price": [4500, 1200, 3500, 28000, 5000, 300, 35000, 6000, 2500, 900]
})

display(dataframe3)
```

```python
# Question 1: Append the records of dataframe1 and dataframe3.

appended_df = pd.concat([dataframe1, dataframe3], ignore_index=True)

display(appended_df)
```

```python
# Question 2: Concatenate the dataframes wherever applicable.

concatenated_df = pd.concat(
    [dataframe1, dataframe2, dataframe3],
    axis=0,
    ignore_index=True,
    sort=False
)

display(concatenated_df)
```

```python
# Question 3: Merge dataframe1 and dataframe2 using ProductID.

merged_df = pd.merge(dataframe1, dataframe2, on="ProductID")

display(merged_df)
```

```python
# Question 4(a): Perform an Inner Join between dataframe1 and dataframe2.

inner_join = pd.merge(dataframe1, dataframe2, on="ProductID", how="inner")

display(inner_join)
```

```python
# Question 4(b): Perform a Left Join between dataframe1 and dataframe2.

left_join = pd.merge(dataframe1, dataframe2, on="ProductID", how="left")

display(left_join)
```

```python
# Question 4(c): Perform a Right Join between dataframe1 and dataframe2.

right_join = pd.merge(dataframe1, dataframe2, on="ProductID", how="right")

display(right_join)
```

```python
# Question 4(d): Perform a Full Outer Join between dataframe1 and dataframe2.

full_outer_join = pd.merge(dataframe1, dataframe2, on="ProductID", how="outer")

display(full_outer_join)
```

```python
# Question: Create a numeric vector using your first name as the object name.
# The first value represents the last four digits of Registration Number 23BDS0257.

Abhishek = np.array([257, 12, 25, 38, 45, 56, 67, 78, 89, 90])

print("Original vector:", Abhishek)
```

```python
# Question: Normalize the data to the range [-3, +3].

normalized_3 = -3 + (
    (Abhishek - Abhishek.min()) /
    (Abhishek.max() - Abhishek.min())
) * 6

print("Normalized data [-3, +3]:", normalized_3)
```

```python
# Question: Normalize the data to the range [-1, +1].

normalized_1 = -1 + (
    (Abhishek - Abhishek.min()) /
    (Abhishek.max() - Abhishek.min())
) * 2

print("Normalized data [-1, +1]:", normalized_1)
```

```python
# Question: Perform Min-Max Normalization to the range [0, 1].

min_max_normalized = (
    (Abhishek - Abhishek.min()) /
    (Abhishek.max() - Abhishek.min())
)

print("Min-Max Normalized data [0, 1]:", min_max_normalized)
```

```python
# Question: Transform the data into Z-score format.

z_score = (Abhishek - Abhishek.mean()) / Abhishek.std(ddof=0)

print("Z-score normalized data:", z_score)
```

```python
# Question: Transform the data using Decimal Scaling Normalization.

j = len(str(int(np.max(np.abs(Abhishek)))))

decimal_scaled = Abhishek / (10 ** j)

print("Decimal Scaling Normalized data:", decimal_scaled)
```

```python
# Question: Display the original vector and all its transformed values.

results = pd.DataFrame({
    "Original": Abhishek,
    "Range [-3, 3]": normalized_3,
    "Range [-1, 1]": normalized_1,
    "Min-Max [0, 1]": min_max_normalized,
    "Z-Score": z_score,
    "Decimal Scaling": decimal_scaled
})

display(results.round(4))
```

```python
# Question 18: Replace missing numerical values with 0 and categorical values with 'Unknown'.

df_constant = df.copy()

for col in df_constant.columns:
    if pd.api.types.is_numeric_dtype(df_constant[col]):
        df_constant[col] = df_constant[col].fillna(0)
    else:
        df_constant[col] = df_constant[col].fillna("Unknown")

display(df_constant.head(10))
```

```python
# Question 1: Import the dataset and load the required Python libraries.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("school93_98.csv")
display(df.head(6))
```

```python
# Question 2: Display the first and last six records of the dataset.

print("First six records:")
display(df.head(6))

print("Last six records:")
display(df.tail(6))
```

```python
# Question 3: Identify the total number of missing values in the dataset.

print("Total missing values:", df.isnull().sum().sum())
```

```python
# Question 4: Display records containing at least one missing value.

display(df[df.isnull().any(axis=1)].head(10))
```

```python
# Question 5: Remove columns containing more than 70% missing values.

missing_percent = df.isnull().mean() * 100
df_clean = df.loc[:, missing_percent <= 70].copy()

print("Columns removed:", missing_percent[missing_percent > 70].index.tolist())
display(df_clean.head(10))
```

```python
# Question 6: Replace missing numerical values with the column mean.

df_mean = df.copy()
numeric_cols = df_mean.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df_mean[col] = df_mean[col].fillna(df_mean[col].mean())

display(df_mean.head(10))
```

```python
# Question 7: Replace missing numerical values with the column median.

df_median = df.copy()
numeric_cols = df_median.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df_median[col] = df_median[col].fillna(df_median[col].median())

display(df_median.head(10))
```

```python
# Question 8: Replace missing categorical values with the mode.

df_mode = df.copy()
categorical_cols = df_mode.select_dtypes(include=["object", "category"]).columns

for col in categorical_cols:
    mode_values = df_mode[col].mode()
    if not mode_values.empty:
        df_mode[col] = df_mode[col].fillna(mode_values.iloc[0])

display(df_mode.head(10))
```

```python
# Question 9: Identify duplicate rows in the dataset.

print("Number of duplicate rows:", df.duplicated().sum())
display(df[df.duplicated(keep=False)].head(10))
```

```python
# Question 10: Remove duplicate rows from the dataset.

df_no_duplicates = df.drop_duplicates().copy()

print("Rows after removing duplicates:", len(df_no_duplicates))
display(df_no_duplicates.head(10))
```

```python
# Question 11: Identify columns whose data types may need correction.

print("Current data types:")
print(df.dtypes)

for col in ["distid", "schid", "year"]:
    if col in df.columns:
        print(f"{col} may be treated as a categorical variable.")
```

```python
# Question 12: Convert character variables into categorical data types.

df_factor = df.copy()
text_cols = df_factor.select_dtypes(include="object").columns

for col in text_cols:
    df_factor[col] = df_factor[col].astype("category")

print(df_factor.dtypes)
```

```python
# Question 13: Convert text values into lowercase and uppercase.

df_text = df.copy()
text_cols = df_text.select_dtypes(include="object").columns

for col in text_cols:
    df_text[col + "_lower"] = df_text[col].str.lower()
    df_text[col + "_upper"] = df_text[col].str.upper()

display(df_text.head(10))
```

```python
# Question 14: Detect outliers in exppp using a boxplot.

sns.boxplot(x=df["exppp"])
plt.title("Boxplot of Expenditure per Pupil")
plt.xlabel("Expenditure per Pupil")
plt.show()
```

```python
# Question 15: Identify outliers in exppp using the IQR method.

Q1 = df["exppp"].quantile(0.25)
Q3 = df["exppp"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df["exppp"] < lower_bound) | (df["exppp"] > upper_bound)]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
display(outliers.head(10))
```

```python
# Question 16: Remove extreme outliers in exppp using the 3 × IQR rule.

Q1 = df["exppp"].quantile(0.25)
Q3 = df["exppp"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 3 * IQR
upper_bound = Q3 + 3 * IQR

df_no_outliers = df[
    df["exppp"].between(lower_bound, upper_bound) |
    df["exppp"].isna()
].copy()

print("Rows before:", len(df))
print("Rows after removing extreme outliers:", len(df_no_outliers))
display(df_no_outliers.head(10))
```

```python
# Question 18: Add a density curve to the histogram of exppp.

sns.histplot(data=df, x="exppp", bins=15, kde=True)
plt.title("Histogram with Density Curve")
plt.show()
```

```python
# Question 1: Load the dataset for bivariate analysis.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

a4_df = pd.read_csv("school93_98.csv")

a4_df["year"] = a4_df["year"].astype("category")
a4_df["schid"] = a4_df["schid"].astype("category")

display(a4_df.head(10))
```

```python
# Question 2(a): Create a contingency table and stacked bar plot using Matplotlib.

a4_table = pd.crosstab(a4_df["year"], a4_df["schid"])
print(a4_table)

a4_table.plot(kind="bar", stacked=True)
plt.title("Stacked Bar Plot: Year and School ID")
plt.xlabel("Year")
plt.ylabel("Frequency")
plt.legend(title="School ID")
plt.tight_layout()
plt.show()
```

```python
# Question 2(b): Create a stacked bar plot using the contingency table.

a4_table.plot(kind="bar", stacked=True)
plt.title("Stacked Bar Plot Using Pandas and Matplotlib")
plt.xlabel("Year")
plt.ylabel("Frequency")
plt.legend(title="School ID")
plt.tight_layout()
plt.show()
```

```python
# Question 3: Compare school frequencies across years using a grouped bar plot.

a4_grouped = pd.crosstab(a4_df["year"], a4_df["schid"])
a4_grouped.plot(kind="bar", stacked=False, figsize=(10, 5))

plt.title("Grouped Bar Plot: Year and School ID")
plt.xlabel("Year")
plt.ylabel("Frequency")
plt.legend(title="School ID")
plt.tight_layout()
plt.show()
```

```python
# Question 4: Analyze the relationship between expenditure and mathematics scores.

sns.scatterplot(data=a4_df, x="exppp", y="math4")
plt.title("Expenditure per Pupil vs Mathematics Score")
plt.xlabel("Expenditure per Pupil")
plt.ylabel("Mathematics Score")
plt.show()
```

```python
# Question 5: Plot average mathematics scores across years.

a4_year_math = a4_df.groupby("year", observed=True)["math4"].mean()

plt.plot(a4_year_math.index.astype(str), a4_year_math.values, marker="o")
plt.title("Average Mathematics Score Across Years")
plt.xlabel("Year")
plt.ylabel("Average Mathematics Score")
plt.grid(True)
plt.show()
```

```python
# Question 6: Generate a heatmap showing relationships among numerical variables.

a4_numeric = a4_df[["lunch", "enrol", "exppp", "math4", "rexpp"]]
a4_corr = a4_numeric.corr()

sns.heatmap(a4_corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of School Variables")
plt.tight_layout()
plt.show()
```

```python
# Question 7: Compare average mathematics scores and lunch values across years.

a4_comparison = a4_df.groupby("year", observed=True)[["math4", "lunch"]].mean()

a4_comparison.plot(kind="bar", figsize=(10, 5))
plt.title("Average Mathematics Score and Lunch Value by Year")
plt.xlabel("Year")
plt.ylabel("Average Value")
plt.legend(title="Variable")
plt.tight_layout()
plt.show()
```

```python
# Question 8: Compare expenditure distributions across different years using density curves.

a4_density_data = a4_df.dropna(subset=["exppp", "year"])

sns.kdeplot(
    data=a4_density_data,
    x="exppp",
    hue="year",
    common_norm=False
)

plt.title("Expenditure Distribution Across Years")
plt.xlabel("Expenditure per Pupil")
plt.ylabel("Density")
plt.show()
```

```python
# Question 9: Compare expenditure per pupil across years using a box plot.

sns.boxplot(data=a4_df, x="year", y="exppp")
plt.title("Expenditure per Pupil Across Years")
plt.xlabel("Year")
plt.ylabel("Expenditure per Pupil")
plt.show()
```

```python
# Question 10: Visualize expenditure distributions across years using a violin plot.

sns.violinplot(data=a4_df, x="year", y="exppp")
plt.title("Violin Plot of Expenditure Across Years")
plt.xlabel("Year")
plt.ylabel("Expenditure per Pupil")
plt.show()
```

```python
# Question 11: Combine a violin plot and box plot to compare expenditure across years.

sns.violinplot(data=a4_df, x="year", y="exppp", inner=None, color="lightblue")
sns.boxplot(
    data=a4_df, x="year", y="exppp",
    width=0.2,
    boxprops={"facecolor": "white"},
    showfliers=False
)

plt.title("Combined Violin and Box Plot")
plt.xlabel("Year")
plt.ylabel("Expenditure per Pupil")
plt.show()
```

```python
# Question 12: Compute correlations and display them using a correlogram.

a4_corr_matrix = a4_df[
    ["lunch", "enrol", "exppp", "math4", "rexpp", "lenrol", "lrexpp"]
].corr()

sns.heatmap(
    a4_corr_matrix,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    fmt=".2f"
)

plt.title("Correlogram of Numerical Variables")
plt.tight_layout()
plt.show()
```

```python
# Question 9: Create a parallel coordinate plot using four numerical variables.
from pandas.plotting import parallel_coordinates

a5_parallel = a5_df[["lunch", "enrol", "exppp", "math4", "year"]].dropna()
a5_parallel = a5_parallel.sample(
    n=min(200, len(a5_parallel)), random_state=42
).copy()

a5_features = ["lunch", "enrol", "exppp", "math4"]

for a5_col in a5_features:
    a5_min = a5_parallel[a5_col].min()
    a5_max = a5_parallel[a5_col].max()
    if a5_max != a5_min:
        a5_parallel[a5_col] = (
            (a5_parallel[a5_col] - a5_min) / (a5_max - a5_min)
        )
    else:
        a5_parallel[a5_col] = 0

a5_parallel["year"] = a5_parallel["year"].astype(str)

plt.figure(figsize=(10, 5))
parallel_coordinates(a5_parallel, class_column="year",
                     cols=a5_features, alpha=0.35)
plt.title("Parallel Coordinates of School Variables by Year")
plt.ylabel("Scaled Value (0 to 1)")
plt.xticks(rotation=15)
plt.legend(title="Year", loc="upper left")
plt.show()
```

# Lab Assessment 6 – Time Series Analysis (Python)

```python
# Question 1: Load the AirPassengers dataset.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.datasets import get_rdataset

a6_data = get_rdataset("AirPassengers", "datasets").data
a6_passengers = pd.Series(
    a6_data["value"].values,
    index=pd.date_range("1949-01-01", periods=len(a6_data), freq="MS"),
    name="Passengers"
)

print("Dataset loaded successfully.")
print("Total observations:", len(a6_passengers))
```

```python
# Question 2: Display the first five and last five observations.
print("First five observations:")
display(a6_passengers.head())

print("Last five observations:")
display(a6_passengers.tail())
```

```python
# Question 3: Display the start and end periods.
print("Start period:", a6_passengers.index[0])
print("End period:", a6_passengers.index[-1])
```

```python
# Question 4: Display the frequency and time-series characteristics.
print("Frequency: Monthly (12 observations per year)")
print("Data type:", type(a6_passengers))
print("Index frequency:", a6_passengers.index.freq)
print("Number of observations:", len(a6_passengers))
print("Time interval:", a6_passengers.index.min(), "to", a6_passengers.index.max())
```

```python
# Question 5: Plot the monthly passenger trend.
plt.figure(figsize=(10, 5))
plt.plot(a6_passengers.index, a6_passengers, linewidth=1.8)
plt.title("Monthly International Airline Passengers (1949–1960)")
plt.xlabel("Year")
plt.ylabel("Number of Passengers (thousands)")
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
# Question 6: Decompose the time series into trend, seasonal, and random components.
from statsmodels.tsa.seasonal import seasonal_decompose

a6_decomposition = seasonal_decompose(
    a6_passengers, model="multiplicative", period=12
)

print("Time-series decomposition completed.")
```

```python
# Question 7: Plot the trend, seasonal, and residual components.
a6_decomposition.plot()
plt.suptitle("AirPassengers Time-Series Decomposition", y=1.02)
plt.gcf().set_size_inches(10, 7)
plt.tight_layout()
plt.show()
```

```python
# Question 8: Plot autocorrelation and partial autocorrelation.
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

plot_acf(a6_passengers, lags=36, ax=axes[0])
axes[0].set_title("Autocorrelation (ACF)")

plot_pacf(a6_passengers, lags=36, method="ywm", ax=axes[1])
axes[1].set_title("Partial Autocorrelation (PACF)")

plt.tight_layout()
plt.show()
```

```python
# Question 9: Fit an ARIMA model to the AirPassengers series.
from statsmodels.tsa.arima.model import ARIMA

a6_model = ARIMA(a6_passengers, order=(2, 1, 2))
a6_fitted = a6_model.fit()

print("ARIMA model fitted successfully.")
```

```python
# Question 10: Display and interpret the fitted ARIMA model.
print(a6_fitted.summary())
print("\nEstimated model parameters:")
print(a6_fitted.params)
```

```python
# Question 11: Forecast passenger totals for the next 12 months.
a6_forecast_result = a6_fitted.get_forecast(steps=12)
a6_forecast = a6_forecast_result.predicted_mean
a6_confidence = a6_forecast_result.conf_int()

print("Forecast for the next 12 months:")
display(pd.DataFrame({
    "Forecast Passengers": a6_forecast,
    "Lower Confidence Limit": a6_confidence.iloc[:, 0],
    "Upper Confidence Limit": a6_confidence.iloc[:, 1]
}))
```

```python
# Question 12: Plot actual and forecasted passenger values.
plt.figure(figsize=(10, 5))

plt.plot(a6_passengers.index, a6_passengers,
         label="Actual Passengers", linewidth=1.8)

plt.plot(a6_forecast.index, a6_forecast,
         label="12-Month Forecast", color="red", linewidth=2)

plt.fill_between(
    a6_forecast.index,
    a6_confidence.iloc[:, 0],
    a6_confidence.iloc[:, 1],
    color="red", alpha=0.15, label="Confidence Interval"
)

plt.title("AirPassengers: Actual Values and 12-Month Forecast")
plt.xlabel("Year")
plt.ylabel("Number of Passengers (thousands)")
plt.legend(loc="upper left")
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

# Assessment 8 – K-Means and Hierarchical Clustering (Python)

```python
# Question 1: Load the dataset.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
from scipy.spatial.distance import pdist

a8_df = pd.read_csv("school93_98.csv")

print("Dataset loaded successfully.")
print("Shape:", a8_df.shape)
```

```python
# Question 2: View the dataset.
display(a8_df.head())
print(a8_df.info())
```

```python
# Question 3: Check missing values in each column.
display(a8_df.isnull().sum().to_frame("Missing Values"))
```

```python
# Question 4: Display the distribution of observations by year.
display(a8_df["year"].value_counts().sort_index().to_frame("Frequency"))
```

```python
# Question 5: Select numerical features for K-Means clustering.
a8_features = [
    "lunch", "enrol", "exppp", "math4",
    "rexpp", "found", "lenrol", "lrexpp", "lavgrexpp"
]

a8_numeric = a8_df[a8_features].replace([np.inf, -np.inf], np.nan)
a8_numeric = a8_numeric.fillna(a8_numeric.median())

print("Selected numerical features:")
print(a8_numeric.columns.tolist())
print("Shape:", a8_numeric.shape)
```

```python
# Question 6: Standardize the numerical features.
a8_sample = a8_numeric.sample(
    n=min(2000, len(a8_numeric)), random_state=42
)

a8_scaler = StandardScaler()
a8_scaled = a8_scaler.fit_transform(a8_sample)

print("Standardized data shape:", a8_scaled.shape)
```

```python
# Question 7: Apply K-Means clustering with two clusters.
a8_kmeans2 = KMeans(n_clusters=2, random_state=42, n_init=10)
a8_labels2 = a8_kmeans2.fit_predict(a8_scaled)
```

```python
# Question 8: Display cluster assignments and cluster sizes.
a8_results = a8_sample.copy()
a8_results["Cluster"] = a8_labels2 + 1

display(a8_results.head(10))
print("Cluster sizes:")
display(a8_results["Cluster"].value_counts().sort_index().to_frame("Samples"))
print("WCSS:", round(a8_kmeans2.inertia_, 2))
```

```python
# Question 9: Display the cluster centers in the original feature scales.
a8_centers = pd.DataFrame(
    a8_scaler.inverse_transform(a8_kmeans2.cluster_centers_),
    columns=a8_features,
    index=["Cluster 1", "Cluster 2"]
)

display(a8_centers.round(2))
```

```python
# Question 10: Calculate within-cluster sum of squares for K from 1 to 10.
a8_wcss = []

for a8_k in range(1, 11):
    a8_model = KMeans(n_clusters=a8_k, random_state=42, n_init=10)
    a8_model.fit(a8_scaled)
    a8_wcss.append(a8_model.inertia_)

print(pd.DataFrame({"K": range(1, 11), "WCSS": a8_wcss}).round(2))
```

```python
# Question 11: Plot WCSS to identify the elbow.
plt.figure(figsize=(7, 4))
plt.plot(range(1, 11), a8_wcss, marker="o")
plt.title("Elbow Method for K-Means")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.xticks(range(1, 11))
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Question 12: Estimate the elbow K and apply the final K-Means model.
a8_k_values = np.arange(1, 11)
a8_points = np.column_stack((a8_k_values, a8_wcss))

a8_start = a8_points[0]
a8_end = a8_points[-1]
a8_line = a8_end - a8_start

a8_distances = np.abs(
    np.cross(a8_line, a8_points - a8_start) / np.linalg.norm(a8_line)
)

a8_best_k = int(a8_k_values[1:-1][np.argmax(a8_distances[1:-1])])
print("Estimated elbow K:", a8_best_k)

a8_final_model = KMeans(n_clusters=a8_best_k, random_state=42, n_init=10)
a8_final_labels = a8_final_model.fit_predict(a8_scaled)

a8_results["Final_Cluster"] = a8_final_labels + 1

print("Final cluster sizes:")
display(a8_results["Final_Cluster"].value_counts().sort_index().to_frame("Samples"))
```

# Lab Assessment 9: Principal Component Analysis (PCA)

```python
# Question 1: Load the dataset and import the required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

df9 = pd.read_csv("school93_98.csv")

print("Dataset loaded successfully")
print("Required Python libraries imported successfully")
print("Dataset shape:", df9.shape)
```

```python
# Question 2: Display the total number of rows and columns

print("Total number of rows:", df9.shape[0])
print("Total number of columns:", df9.shape[1])
```

```python
# Question 3: Extract numeric columns, clean data, and normalize it

data_numeric9 = df9.select_dtypes(include=[np.number]).copy()

data_numeric9 = data_numeric9.replace([np.inf, -np.inf], np.nan)
data_numeric9 = data_numeric9.dropna(axis=1, how="all")
data_numeric9 = data_numeric9.fillna(data_numeric9.median())

scaler9 = StandardScaler()
data_scaled = pd.DataFrame(
    scaler9.fit_transform(data_numeric9),
    columns=data_numeric9.columns,
    index=data_numeric9.index
)

print("Scaled data:")
display(data_scaled.head())
```

```python
# Question 4: Perform PCA and print its summary

pca9 = PCA()
pca_result9 = pca9.fit_transform(data_scaled)

pca_summary9 = pd.DataFrame({
    "Principal Component": [
        f"PC{i+1}" for i in range(len(pca9.explained_variance_ratio_))
    ],
    "Standard Deviation": np.sqrt(pca9.explained_variance_),
    "Proportion of Variance": pca9.explained_variance_ratio_,
    "Cumulative Proportion": np.cumsum(pca9.explained_variance_ratio_)
})

display(pca_summary9.round(4))
```

```python
# Question 5: Plot the scree plot with percentage labels

variance9 = pca9.explained_variance_ratio_ * 100
components9 = np.arange(1, len(variance9) + 1)

plt.figure(figsize=(12, 6))
plt.plot(components9, variance9, marker="o", linewidth=2)
plt.xticks(components9)
plt.ylim(0, 60)
plt.xlabel("Principal Components")
plt.ylabel("Percentage of Variance Explained")
plt.title("Scree Plot - Principal Component Analysis")
plt.grid(alpha=0.3)

for x, y in zip(components9, variance9):
    if y <= 60:
        plt.text(x, y + 1, f"{y:.1f}%", ha="center", fontsize=8)

plt.tight_layout()
plt.show()
```
